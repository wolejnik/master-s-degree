```python
import whitebox
import matplotlib.pyplot as plt
import imageio
import laspy as lp
import pylas 
import os
import numpy as np
from mpl_toolkits.mplot3d import Axes3D
import matplotlib.pyplot as plt


def Zadanie1():
    # init
    wbt = whitebox.WhiteboxTools()
    data_dir = "C:\\Users\\skork\\Desktop\\Semestr 2\\Języki Analizy Danych\\A.Lysko\\Lab3"
    wbt.set_working_dir(data_dir)
    wbt.verbose = False
    las = "swidwie_dense_cloud_crop.las"
    resolution = 0.1
    outFile = "las_grid.tif"
    # funkcje whiteboxtool https://rdrr.io/rforge/whitebox/man/lidar_nearest_neighbour_gridding.html
    # interpolacja
    wbt.lidar_nearest_neighbour_gridding(las, outFile, parameter = "elevation", returns = "all", resolution = resolution,
    radius = 2.5)
    # wyświetlanie
    raster = imageio.imread(os.path.join(data_dir, outFile))
    plt.imshow(raster)
    plt.title('Nearest - resampling DEM')
    plt.show()

def Zadanie2():
    laz_filename = 'szczecin.laz'
    las_filename = 'szczecin.las'
    # konwersja z .laz do .las
    las = pylas.read(laz_filename)    
    las = pylas.convert(las)    
    las.write(las_filename)
    # init
    wbt = whitebox.WhiteboxTools()
    data_dir = "C:\\Users\\skork\\Desktop\\Semestr 2\\Języki Analizy Danych\\A.Lysko\\Lab3"
    wbt.set_working_dir(data_dir)
    # Poranie informacji o pliku
    wbt.lidar_info(las_filename)
    wbt.verbose = False
    outFile = "szczecinClassified.las"
    # Wybór punktów w klasach
    wbt.filter_lidar_classes(las_filename, outFile, exclude_cls = "0,1,7,8,10,11,12,13,14,15,16,17,18")
    # usuwanie szumów
    wbt.lidar_remove_outliers(outFile, outFile, radius = 1, elev_diff = 1, use_median = 1, classify = 1,)
    gridFile = "szczecinClassified.tif"
    # Konwersja do grida
    wbt.lidar_nearest_neighbour_gridding(outFile, gridFile, parameter = "elevation", returns = "all", resolution = 0.1, radius = 0.1)
    # Wyświetlenie rezultatu
    raster = imageio.imread(os.path.join(data_dir, gridFile))
    plt.figure()
    plt.imshow(raster)
    plt.legend()
    plt.show()

def Zadanie3():
    pass
if __name__ == "__main__":
    # Zadanie1()
    Zadanie2()
    Zadanie3()
```
