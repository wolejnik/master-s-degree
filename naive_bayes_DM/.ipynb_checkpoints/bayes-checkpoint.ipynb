```python
import numpy as np
from sklearn.base import BaseEstimator, ClassifierMixin
from sklearn import metrics
import matplotlib.pyplot as plt
```

```python
def extract_attr_count(attr, uniques_ext):
    """
   Take a 1D numpy array as input and return a dict mapping values to counts
   """
    uniques, count = np.unique(attr, return_counts=True)
    counts = {u: count[i] for i, u in enumerate(uniques)}
    for unique in uniques_ext:
        if unique not in counts:
            counts[unique] = 0
    return counts

class NaiveBayesClassifier(BaseEstimator, ClassifierMixin):
    alpha = 0.01

    def fit(self, X, d):
        # liczba dokumentow w D
        self.cls, self.n_cls = np.unique(d, return_counts=True)
        self.classes_ = self.cls
        # logprior[c]
        self.p_cls = np.log(self.n_cls / float(np.sum(self.n_cls)))
        self.counts = {elm: None for elm in self.cls}
        self.n_attrs = X.T.shape[0]
        uniques = [np.unique(attr) for attr in X.T]
        
        # dla kazdej z klas obliczenie P(c) termns
        for elm in self.cls:
            self.counts[elm] = [extract_attr_count(L, uniques[i]) for i, L in enumerate(X[d == elm].T)]

        return self

    def predict_log_proba(self, X):
        preds = np.zeros((X.shape[0], len(self.cls)))
        # dla kazdego dokumetu
        for num, xi in enumerate(X):
            Pxi = []
            # dla kazdej z klas 
            for i, cls in enumerate(self.cls):
                tmp = []
                # czy zawiera sie w słowaniku V
                for j, v in enumerate(xi):
                    try:
                        attr_count = self.counts[cls][j].get(v, 0)
                        attrs_len = len(self.counts[cls][j])
                        # Compute probabilities with additive smoothing
                        # dodawanie małej wartości do prawdopodobieństwa danego punktu, żeby potem całość się nie zerowała
                        tmp.append((attr_count + self.alpha) / ((self.n_cls[i] + self.alpha) * attrs_len)) # including alpha avoids log(0) and bumps accuracy by 0.2
                    except IndexError: #word from test data wasn't in train data
                        continue
                Pxi.append(tmp)
            preds[num] = np.array([self.p_cls[i] + np.sum(np.log(Pxi[i])) for i in range(len(self.cls))])
        return preds

    def predict(self, X):
        log_preds = self.predict_log_proba(X)

        return self.cls[np.argmax(log_preds, axis=1)]

    def predict_proba(self, X):
        return np.exp(self.predict_log_proba(X))
```

```python
def prepare_data(filename):
    max_column = 0
    max_row = 0
    with open(filename, 'r') as data:
        max_len = data[0].max()
        print(f'max_len => {max_len}')
        rows = data.readlines()

    for row in rows:
        row, col, value = row.split(' ')
        row = int(row)
        if row > max_row:
            max_row = row
        col = int(col)
        if col > max_column:
            max_column = col

    X = np.zeros((max_row, max_column))
    for number, row in enumerate(rows):
        row, col, value = row.split(' ')
        X[int(row)-1][int(col)-1] = int(value)

    return X
```
